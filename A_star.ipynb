```python
# A* Search

import heapq


def build_graph():
    graph = {}
    n = int(input("Enter number of nodes: "))
    print("Enter node names one by one:")
    nodes = []
    for i in range(n):
        node = input(f"  Node {i + 1}: ").strip()
        nodes.append(node)
        graph[node] = []

    e = int(input("Enter number of edges: "))
    print("Enter each edge as: from_node to_node cost")
    undirected = input("Is the graph undirected? (y/n): ").strip().lower() == 'y'

    for i in range(e):
        while True:
            try:
                parts = input(f"  Edge {i + 1} (from to cost): ").split()
                u, v, cost = parts[0], parts[1], float(parts[2])
                break
            except (ValueError, IndexError):
                print("  Invalid format. Example: A B 4")
        graph[u].append((v, cost))
        if undirected:
            graph[v].append((u, cost))

    return graph, nodes


def get_heuristics(nodes):
    print("\nEnter heuristic value h(n) for each node (estimated cost to goal):")
    h = {}
    for node in nodes:
        while True:
            try:
                h[node] = float(input(f"  h({node}): "))
                break
            except ValueError:
                print("  Please enter a number.")
    return h


def reconstruct_path(came_from, current):
    path = [current]
    while current in came_from:
        current = came_from[current]
        path.append(current)
    path.reverse()
    return path


def a_star_search(graph, start, goal, h):
    frontier = [(h.get(start, 0), 0, start)]   # (f = g + h, g, node)
    came_from = {}
    g_score = {start: 0}
    visited_order = []

    while frontier:
        f, g, current = heapq.heappop(frontier)

        if current in visited_order:
            continue
        visited_order.append(current)

        if current == goal:
            return reconstruct_path(came_from, current), g, visited_order

        for neighbor, edge_cost in graph.get(current, []):
            tentative_g = g + edge_cost
            if neighbor not in g_score or tentative_g < g_score[neighbor]:
                g_score[neighbor] = tentative_g
                came_from[neighbor] = current
                f_score = tentative_g + h.get(neighbor, 0)
                heapq.heappush(frontier, (f_score, tentative_g, neighbor))

    return None, float('inf'), visited_order


def main():
    print("=== Graph Setup ===")
    graph, nodes = build_graph()

    start = input("\nEnter start node: ").strip()
    goal = input("Enter goal node: ").strip()

    h = get_heuristics(nodes)

    path, cost, order = a_star_search(graph, start, goal, h)

    print("\n--- A* Search Result ---")
    if path:
        print("Path found :", " -> ".join(path))
        print("Total cost :", cost)
    else:
        print("No path found.")
    print("Nodes explored (order):", " -> ".join(order))


if __name__ == "__main__":
    main()
```
