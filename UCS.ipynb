```python
# Uniform Cost Search (UCS)

import heapq


def build_graph():
    graph = {}
    n = int(input("Enter number of nodes: "))
    print("Enter node names one by one:")
    for i in range(n):
        node = input(f"  Node {i + 1}: ").strip()
        graph[node] = []

    e = int(input("Enter number of edges: "))
    print("Enter each edge as: from_node to_node cost")
    undirected = input("Is the graph undirected? (y/n): ").strip().lower() == 'y'

    for i in range(e):
        while True:
            try:
                parts = input(f"  Edge {i + 1} (from to cost): ").split()
                u, v, cost = parts[0], parts[1], float(parts[2])
                break
            except (ValueError, IndexError):
                print("  Invalid format. Example: A B 4")
        graph[u].append((v, cost))
        if undirected:
            graph[v].append((u, cost))

    return graph


def reconstruct_path(came_from, current):
    path = [current]
    while current in came_from:
        current = came_from[current]
        path.append(current)
    path.reverse()
    return path


def uniform_cost_search(graph, start, goal):
    frontier = [(0, start)]          # (path_cost, node)
    came_from = {}
    cost_so_far = {start: 0}
    visited_order = []

    while frontier:
        cost, current = heapq.heappop(frontier)

        if current in visited_order:
            continue
        visited_order.append(current)

        if current == goal:
            return reconstruct_path(came_from, current), cost, visited_order

        for neighbor, edge_cost in graph.get(current, []):
            new_cost = cost + edge_cost
            if neighbor not in cost_so_far or new_cost < cost_so_far[neighbor]:
                cost_so_far[neighbor] = new_cost
                came_from[neighbor] = current
                heapq.heappush(frontier, (new_cost, neighbor))

    return None, float('inf'), visited_order


def main():
    print("=== Graph Setup ===")
    graph = build_graph()

    start = input("\nEnter start node: ").strip()
    goal = input("Enter goal node: ").strip()

    path, cost, order = uniform_cost_search(graph, start, goal)

    print("\n--- Uniform Cost Search Result ---")
    if path:
        print("Path found :", " -> ".join(path))
        print("Total cost :", cost)
    else:
        print("No path found.")
    print("Nodes explored (order):", " -> ".join(order))


if __name__ == "__main__":
    main()
```
