### K-Means Clustering Results
Here, we fit the K-Means clustering model to our dataset and display the resulting cluster labels, cluster centroids, and the final visualization.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans

# 1. Generate a synthetic 2D dataset
X, y_true = make_blobs(n_samples=500, centers=4, cluster_std=0.60, random_state=42)

# 2. Train K-Means model
optimal_k = 4
kmeans = KMeans(n_clusters=optimal_k, init='k-means++', max_iter=300, n_init=10, random_state=42)
y_kmeans = kmeans.fit_predict(X)
centroids = kmeans.cluster_centers_

# Print cluster labels and centroid coordinates
print("Assigned Cluster Labels (first 30 points):")
print(y_kmeans[:30])
print("\nCluster Centroids Coordinates:")
for i, centroid in enumerate(centroids):
    print(f"Cluster {i}: Centroid = {centroid}")
```

```python
# 3. Visualize the clustered data
plt.figure(figsize=(8, 6))
colors = ['purple', 'teal', 'green', 'gold']
for idx in range(optimal_k):
    plt.scatter(X[y_kmeans == idx, 0], X[y_kmeans == idx, 1], s=30, c=colors[idx], label=f'Cluster {idx}')

# Plot the centroids
plt.scatter(centroids[:, 0], centroids[:, 1], s=200, c='red', marker='X', edgecolors='black', label='Centroids')

plt.title('K-Means Clustering Results')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.legend()
plt.grid(True)
plt.show()
```
